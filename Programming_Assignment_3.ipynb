```python
import pandas as pd #This imports the Panda Library and uses the "pd" keyword to shorten the codings.
```

```python
cars = pd.read_csv('cars.csv') #Reads the csv files and stores it into a DataFrame to be use throughout.
cars #Displays the contents of the DataFrame.
```

**A.  POSITIONAL AND LABEL-BASED SLICING**

*A.2. Using "iloc", show cars_6_to_10.*

```python
cars_6_to_10 = cars.iloc[5:10] #Specified in the problem that using "iloc" select only the rows 6 to 10, starting from zero (0) as the first index.
cars_6_to_10 #Displays the rows 6 to 10.
```

*A.3. Using "column labels", display cars_6_to_10 specific columns.*

```python
cars_6_to_10[['Model', 'mpg', 'cyl', 'hp', 'gear']] #Shows only the given required columns.
```

**B. MODEL LOOKUP**

*B.1. Display the complete row for Toyota Corolla and store it in Toyota.*

```python
Toyota = cars.loc[cars['Model'] == 'Toyota Corolla'] #Locates the row where Toyota Corolla is.
Toyota #Shows all the contents of Toyota Corolla's Row.
```

*B.2. Display the Model, mpg, hp and wt columns of Pontaic Firebird and store it in Pontaic.*

```python
Pontiac = cars.loc[cars['Model'] == 'Pontiac Firebird'] #Locate only the row for the Pontiac Firebird.
Pontiac = Pontiac [['Model', 'mpg', 'hp', 'wt']] #Shows only the selected columns.
Pontiac #Displays the chosen values.
```

**C. MULTI-MODEL SUBSETTING**

```python
selected_cars = pd.DataFrame(cars[(cars ['Model'] == 'Datsun 710') | (cars ['Model'] == 'Lotus Europa') | (cars ['Model'] == 'Ferrari Dino')], #Creates a new DataFrame called "selected_cars" and selects only the Three required Models.
                             columns = ['Model', 'mpg', 'cyl', 'hp', 'gear']) #Chooses only the selected columns.
selected_cars #Displays the selected cars.
```
